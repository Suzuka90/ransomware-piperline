```python
import pandas as pd
import numpy as np
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, accuracy_score
from sklearn.preprocessing import StandardScaler
import pickle

print("🔬 RANSOMWARE ML PIPELINE - ULTRA FIXED")
print("=" * 60)

class MalwarePipeline:
    def __init__(self):
        self.model = None
        self.scaler = StandardScaler()
    
    def create_dataset(self):
        """✅ Dataset FIXED - NO np.random.choice ERROR"""
        np.random.seed(42)
        n_samples = 500
        
        # Labels corretti (lista, non dict.values())
        labels = np.repeat([0,1,2,3], [200, 100, 100, 100])  # Bilanciati
        
        data = {
            'file_size_kb': np.random.lognormal(6.5, 1.2, n_samples),
            'entropy': np.random.normal(6.8, 0.8, n_samples),
            'pe_imports': np.random.poisson(35, n_samples),
            'sections_count': np.random.randint(2, 12, n_samples),
            'has_upx': np.random.binomial(1, 0.3, n_samples),
            'suspicious_strings': np.random.poisson(5, n_samples),
            'label': labels
        }
        
        df = pd.DataFrame(data)
        
        # Feature engineering per classi
        df.loc[df['label'] == 2, 'file_size_kb'] *= 3.0   # WannaCry grande
        df.loc[df['label'] == 3, 'entropy'] += 0.8       # NotPetya alta entropia
        df.loc[df['label'] == 1, 'pe_imports'] *= 1.5    # CryptoLocker imports
        
        print(f"✅ Dataset creato: {len(df)} campioni")
        return df
    
    def train(self):
        """✅ TRAINING 100% FUNZIONANTE"""
        print("\n📊 Preprocessing...")
        df = self.create_dataset()
        
        features = ['file_size_kb', 'entropy', 'pe_imports', 'sections_count', 'has_upx', 'suspicious_strings']
        X = df[features].values
        y = df['label'].values
        
        # Split
        X_train, X_test, y_train, y_test = train_test_split(
            X, y, test_size=0.25, random_state=42, stratify=y
        )
        
        # Scale
        X_train_s = self.scaler.fit_transform(X_train)
        X_test_s = self.scaler.transform(X_test)
        
        # Model
        print("🤖 Addestramento Random Forest...")
        self.model = RandomForestClassifier(
            n_estimators=150, max_depth=10, random_state=42, n_jobs=-1
        )
        self.model.fit(X_train_s, y_train)
        
        # Results
        train_acc = accuracy_score(y_train, self.model.predict(X_train_s))
        test_acc = accuracy_score(y_test, self.model.predict(X_test_s))
        
        print(f"\n📈 RISULTATI:")
        print(f"Train Accuracy: {train_acc:.3f}")
        print(f"Test Accuracy:  {test_acc:.3f}")
        
        y_pred = self.model.predict(X_test_s)
        print("\n📊 REPORT:")
        print(classification_report(y_test, y_pred, 
                  target_names=['Benigno','CryptoLocker','WannaCry','NotPetya']))
        
        return test_acc
    
    def predict_demo(self):
        """Demo predizione"""
        # Simulazione WannaCry
        sample = np.array([[7500, 8.2, 70, 8, 1, 15]])  # Caratteristiche WannaCry
        sample_scaled = self.scaler.transform(sample)
        
        pred = self.model.predict(sample_scaled)[0]
        probs = self.model.predict_proba(sample_scaled)[0]
        
        names = ['Benigno','CryptoLocker','WannaCry','NotPetya']
        print(f"\n🧪 PREDIZIONE DEMO:")
        print(f"Predizione: {names[pred]}")
        print(f"Confidence: {np.max(probs):.1%}")
        print("Probabilità:", {names[i]: f"{p:.1%}" for i,p in enumerate(probs)})
    
    def save_model(self):
        """Salva modello"""
        with open('ransomware_final_model.pkl', 'wb') as f:
            pickle.dump({
                'model': self.model,
                'scaler': self.scaler,
                'features': ['file_size_kb', 'entropy', 'pe_imports', 'sections_count', 'has_upx', 'suspicious_strings']
            }, f)
        print("💾 Modello salvato!")

def main():
    """MAIN SEMPLICE E FUNZIONANTE"""
    pipeline = MalwarePipeline()
    
    accuracy = pipeline.train()
    pipeline.predict_demo()
    pipeline.save_model()
    
    print(f"\n🎉 PIPELINE COMPLETATA! Accuracy: {accuracy:.3f}")

if __name__ == "__main__":
    main()
```
